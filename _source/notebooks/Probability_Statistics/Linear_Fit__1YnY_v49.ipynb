```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-10,10,20)
```

```python
x
```

```python
m = 50
b = 17 
```

```python
y = m * x + b
```

```python
plt.plot(x,y,'.')
```

```python
r = 100*np.random.randn(20,)
```

```python
r
```

```python
plt.plot(r,'.')
```

```python
y2 = m * x + b + r
```

```python
plt.plot(x,y2,'.')
```

```python
m_guess = 1
b_guess = 1

y_guess = m_guess * x + b_guess

plt.plot(x,y_guess,'r.');
plt.plot(x,y2,'g.');
```

```python
for m_guess in np.linspace(-100,100,100):
    
        y_guess = m_guess * x 

        plt.plot(x,y_guess,'r.');
        plt.plot(x,y2,'g.');
        plt.show()

        loss = np.sum((y_guess - y)**2)

        print(loss,'M = ',m_guess)
```

```python
best_loss = 100000000000
best_m = 0
best_b = 0 

for m_guess in np.linspace(-100,100,100):
    for b_guess in np.linspace(-100,100,100):
        
        y_guess = m_guess * x + b_guess

        loss = np.sum((y_guess - y)**2)

        if loss < best_loss:
            
            best_loss = loss
            best_m = m_guess
            best_b = b_guess 
```

```python
y_guess = best_m * x + best_b

plt.plot(x,y_guess,'r.');
plt.plot(x,y2,'g.');
plt.show()

print(best_loss,best_m,best_b)
```
