```python
# Step 1: Clone the GitHub repo
!git clone https://github.com/williamedwardhahn/MathData25.git

# Step 2: Import libraries
import os
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# Step 3: Set up paths
data_dir = '/content/MathData25/toast'
train_dir = os.path.join(data_dir, 'train')
test_dir = os.path.join(data_dir, 'test')

# Step 4: Define image transforms
transform = transforms.Compose([
    transforms.Resize((64, 64)),
    transforms.ToTensor()
])

# Step 5: Load datasets
train_dataset = datasets.ImageFolder(root=train_dir, transform=transform)
test_dataset = datasets.ImageFolder(root=test_dir, transform=transform)

# Step 6: Create data loaders (optional)
train_loader = DataLoader(train_dataset, batch_size=32, shuffle=True)
test_loader = DataLoader(test_dataset, batch_size=32, shuffle=False)

# Step 7: Convert entire datasets into tensors
train_images = torch.stack([img for img, label in train_dataset])
train_labels = torch.tensor([label for img, label in train_dataset])
test_images = torch.stack([img for img, label in test_dataset])
test_labels = torch.tensor([label for img, label in test_dataset])

# Step 8: Print class labels
print("Classes:", train_dataset.classes)
print("Class to index mapping:", train_dataset.class_to_idx)
print("Train images shape:", train_images.shape)
print("Train labels shape:", train_labels.shape)
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Function to show a montage of images with labels
def show_montage(images, labels, class_names, rows=4, cols=8):
    fig, axes = plt.subplots(rows, cols, figsize=(cols * 1.5, rows * 1.5))
    for i in range(rows * cols):
        if i >= len(images):
            break
        img = images[i].permute(1, 2, 0).numpy()  # CHW to HWC
        ax = axes[i // cols, i % cols]
        ax.imshow(img)
        ax.set_title(class_names[labels[i]])
        ax.axis('off')
    plt.tight_layout()
    plt.show()

# Plot montage of first N training images
show_montage(train_images, train_labels, train_dataset.classes, rows=4, cols=8)
```
