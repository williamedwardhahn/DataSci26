```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from io import BytesIO
import requests

# Load image from URL
response = requests.get(url)
image = Image.open(BytesIO(response.content))
gray_image = image.convert('L')
rgb_image = image.convert('RGB')

def load_image_from_url(url):
    """Load an image from a given URL."""
    response = requests.get(url)
    return Image.open(BytesIO(response.content))

def display_image(array_data, title, cmap=None):
    """Display the image with a given title."""
    img = Image.fromarray(array_data.astype(np.uint8))
    plt.figure(figsize=(5, 5))
    plt.imshow(img, cmap=cmap)
    plt.title(title)
    plt.axis('off')
    plt.tight_layout()
    plt.show()

def reduce_bit_depth(img, bit_depth):
    """Reduce the bit depth of an image."""
    factor = 256 // (2 ** bit_depth)
    reduced_img = (np.array(img) // factor) * factor
    return reduced_img
```

```python
# Display original, grayscale, and RGB images
display_image(np.array(image), 'Original Image')
display_image(np.array(gray_image), 'Grayscale (8-bit)', 'gray')
display_image(np.array(rgb_image), 'RGB (8-bit)')
```

```python
def generate_grayscale_palette(bit_depth):
    """Generate a grayscale palette for a given bit depth."""
    num_shades = 2 ** bit_depth
    values = np.linspace(0, 255, num_shades, dtype=np.uint8)
    palette = np.zeros((num_shades, num_shades), dtype=np.uint8)
    for i in range(num_shades):
        palette[:, i] = values[i]
    return palette

# Display grayscale palettes for different bit depths
bit_depths = [1, 2, 3, 5, 6, 7, 8]
for depth in bit_depths:
    palette = generate_grayscale_palette(depth)
    plt.imshow(palette, cmap='gray')
    num_shades = 2 ** depth
    plt.title(f'Grayscale ({depth}-bit)  Number of Shades: {num_shades}')
    plt.axis('off')
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def generate_rgb_palette(bit_depth):
    """Generate an RGB palette for a given bit depth."""
    num_shades = 2 ** bit_depth
    values = np.linspace(0, 255, num_shades, dtype=np.uint8)

    total_colors = num_shades ** 3
    side_length = int(np.ceil(np.sqrt(total_colors)))

    palette = np.zeros((side_length, side_length, 3), dtype=np.uint8)

    counter = 0
    for red in values:  # Modified order
        for green in values:
            for blue in values:
                row = counter // side_length
                col = counter % side_length
                palette[row, col] = [red, green, blue]
                counter += 1

    return palette

# Display RGB palettes for 1-bit, 2-bit, and 3-bit
for depth in [1, 2, 3, 4, 5]:
    rgb_palette = generate_rgb_palette(depth)
    plt.imshow(rgb_palette)
    plt.title(f'RGB ({depth}-bit) Palette')
    plt.axis('off')
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import colorsys

def generate_rgb_palette(bit_depth):
    """Generate an RGB palette sorted by hue for a given bit depth."""
    num_shades = 2 ** bit_depth
    values = np.linspace(0, 255, num_shades, dtype=np.uint8)

    # Generate all RGB colors
    colors = []
    for red in values:
        for green in values:
            for blue in values:
                colors.append([red, green, blue])

    # Convert each RGB color to HSV and sort by hue, saturation, then value
    colors = sorted(colors, key=lambda rgb: colorsys.rgb_to_hsv(rgb[0]/255, rgb[1]/255, rgb[2]/255))

    total_colors = len(colors)
    side_length = int(np.ceil(np.sqrt(total_colors)))
    palette = np.zeros((side_length, side_length, 3), dtype=np.uint8)

    counter = 0
    for color in colors:
        row = counter // side_length
        col = counter % side_length
        palette[row, col] = color
        counter += 1

    return palette

# Display RGB palettes for 1-bit, 2-bit, 3-bit, 4-bit, and 5-bit
for depth in [1, 2, 3, 4, 5]:
    rgb_palette = generate_rgb_palette(depth)
    plt.imshow(rgb_palette)
    plt.title(f'RGB ({depth}-bit) Palette')
    plt.axis('off')
    plt.show()
```

### 2.1 Binary Images

A binary image is one where each pixel can be either black (0) or white (1). If you have an image of size $m \times n$, then it's represented as an $m \times n$ matrix with each entry being 0 or 1.

```python
2**9 #Binary 3x3
```



### 2.2 Grayscale Images

Grayscale images have more depth than binary images. Each pixel in a grayscale image can take a value between 0 (black) and 255 (white). Each pixel in a grayscale image is typically represented using 8 bits.

```python
256**9 #Grayscale 3x3
```



### 2.3 RGB Images

Color images are usually represented in the RGB format. Each pixel has three values corresponding to the Red, Green, and Blue channels. Each channel can take a value between 0 and 255. Therefore, a pixel in an RGB image requires $$3 \times 8 = 24$$ bits.



## 3. Combinatorics of Images

Combinatorics is the branch of mathematics dealing with combinations of objects in specific sets under certain constraints.

### 3.1 Binary Images

For a binary image of size $m \times n$, each pixel has 2 possible values. So, the number of possible images is $$2^{m \times n}$$.

### 3.2 Grayscale Images

For a grayscale image of size $m \times n$, each pixel can take 256 values. Hence, the number of possible images is $$256^{m \times n}$$.

### 3.3 RGB Images

For an RGB image of size $m \times n$, each pixel can take $256 \times 256 \times 256$ values. Thus, the number of possible images is $$(256 \times 256 \times 256)^{m \times n}$$.

## 4. Hyperastronomical Numbers

The combinatorics of images lead to very large numbers, especially for bigger images. Numbers larger than the number of atoms in the observable universe (estimated at about $10^{80}$) are sometimes referred to as "hyperastronomical." The vast number of possible images, even for small dimensions, often falls into this category.

For a small $3 \times 3$ image there are approximately:
- **Binary Images**: $$(2)^9 = 512$$
- **Grayscale Images**:  $$(256)^9 = 4.72 \times 10^{18}$$
- **RGB Images**: $$(256^3)^9 = 1.05 \times 10^{44}$$

possible images.

The combinatorics behind images underscores the vastness of the possible image space. This vast space poses challenges and opportunities for data scientists. Algorithms need to generalize well across this space based on a small sample of actual images, and understanding the underlying math helps in refining these algorithms.


```python
# Define function to calculate number of possibilities
def calculate_possibilities(image_size):
    binary_possibilities = 2**(image_size * image_size)
    grayscale_possibilities = 256**(image_size * image_size)
    rgb_possibilities = (256 * 256 * 256)**(image_size * image_size)
    return binary_possibilities, grayscale_possibilities, rgb_possibilities





# Define range of image sizes
image_sizes = list(range(1, 5))

# Calculate possibilities for each image size
binary_results, grayscale_results, rgb_results = [], [], []

for size in image_sizes:

    binary, grayscale, rgb = calculate_possibilities(size)

    print(f'Binary possibilities for size {size}x{size} = {binary}')
    print(f'Grayscale possibilities for size {size}x{size} = {grayscale}')
    print(f'RGB possibilities for size {size}x{size} = {rgb}')
    print("****************************************************")

    binary_results.append(binary)
    grayscale_results.append(grayscale)
    rgb_results.append(rgb)
```

```python
# Plot the results
plt.figure(figsize=(12, 6))
plt.plot(image_sizes, binary_results, '-o', label="Binary Images")
plt.plot(image_sizes, grayscale_results, '-o', label="Grayscale Images")
plt.plot(image_sizes, rgb_results, '-o', label="RGB Images (upto 5x5)")
plt.yscale("log")
plt.xlabel("Image Size (width of square image)")
plt.ylabel("Number of Possible Images (log scale)")
plt.title("Growth of Number of Possible Images with Image Size")
plt.legend()
plt.grid(True, which="both", linestyle="--", linewidth=0.5)
plt.tight_layout()
plt.show()
```

```python
# @title Binary

import matplotlib.pyplot as plt
import matplotlib.patches as patches

def plot_binary(num):
    # Convert number to binary and pad with zeros to make it 4 bits long
    bin_str = format(num, '04b')

    # Create a new figure and axis
    fig, ax = plt.subplots(figsize=(5, 0.7))
    ax.set_xlim(0, 4)
    ax.set_ylim(0, 1)

    # Remove axis labels and ticks
    ax.axis('off')

    # For each bit in the binary string, plot a filled or empty square
    for i, bit in enumerate(bin_str):
        color = 'gray' if bit == '1' else 'white'
        rect = patches.Rectangle((i, 0), 1, 1, facecolor=color, edgecolor='blue', linewidth=1.5)
        ax.add_patch(rect)


    # Add decimal representation to the title
    ax.set_title(f" Decimal: {num} = Binary: {bin_str}")

    # Show the plot
    plt.show()
```

```python
# Plot binary representations for numbers 0 to 15
for i in range(16):
    plot_binary(i)
```

```python
# @title
import numpy as np
import matplotlib.pyplot as plt

def generate_binary_matrices(n):
    """Generates all possible binary matrices of size n x n."""
    num_matrices = 2**(n*n)
    matrices = []

    for i in range(num_matrices):
        binary = format(i, f'0{n*n}b')  # Convert i to binary with fixed width n*n
        matrix = np.array([int(bit) for bit in binary]).reshape(n, n)
        matrices.append(matrix)

    return matrices

def display_matrices(matrices, title):
    """Displays a list of matrices using matplotlib."""
    n = len(matrices)

    fig, axs = plt.subplots(1, n, figsize=(n*2, 2))

    for i, matrix in enumerate(matrices):
        axs[i].imshow(matrix, cmap="gray")
        axs[i].axis('off')

    plt.suptitle(title)
    plt.tight_layout()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

def recursive_average(colors):
    """
    Compute the recursive average of the given list of colors.

    Parameters:
    - colors: List of color arrays

    Returns:
    - A new list of color arrays containing the recursive averages.
    """
    new_colors = [colors[0]]
    for i in range(1, len(colors)):
        avg_color = (colors[i] + new_colors[-1]) / 2.0
        new_colors.extend([avg_color, colors[i]])
    return new_colors

def generate_recursive_colors_v2(passes):
    """
    Generate a visualization of 2 random colors and their recursive averages for a number of passes.

    Parameters:
    - passes: Number of times to apply the recursive averaging process

    Returns:
    - An image showing the 2 endpoint colors and their recursive averages.
    """
    # Start with 2 random colors
    colors = [np.ones((10,10,3)) for _ in range(2)]
    for color in colors:
        color[:,:,0] = np.random.rand()
        color[:,:,1] = np.random.rand()
        color[:,:,2] = np.random.rand()

    # Apply the recursive averaging for the specified number of passes
    for _ in range(passes):
        colors = recursive_average(colors)

    # Horizontally stack the colors
    result = np.hstack(colors)

    return result
```

```python
import numpy as np
import matplotlib.pyplot as plt

def recursive_average(colors):
    """
    Compute the recursive average of the given list of colors.

    Parameters:
    - colors: List of color arrays

    Returns:
    - A new list of color arrays containing the recursive averages.
    """
    new_colors = [colors[0]]
    for i in range(1, len(colors)):
        avg_color = (colors[i] + new_colors[-1]) / 2.0
        new_colors.extend([avg_color, colors[i]])
    return new_colors

def generate_recursive_colors_v2(passes, initial_colors):
    """
    Generate a visualization of 2 given colors and their recursive averages for a number of passes.

    Parameters:
    - passes: Number of times to apply the recursive averaging process
    - initial_colors: List of initial colors to start with

    Returns:
    - An image showing the 2 endpoint colors and their recursive averages.
    """
    colors = initial_colors.copy()

    # Apply the recursive averaging for the specified number of passes
    for _ in range(passes):
        colors = recursive_average(colors)

    # Horizontally stack the colors
    result = np.hstack(colors)

    return result

# Generate the same initial random colors for all passes
initial_colors = [np.ones((10, 10, 3)) for _ in range(2)]
for color in initial_colors:
    color[:,:,0] = np.random.rand()
    color[:,:,1] = np.random.rand()
    color[:,:,2] = np.random.rand()

# Generate and display for 3 and 4 passes
passes_values = [0,1,2,3,4]
imgs = [generate_recursive_colors_v2(p, initial_colors) for p in passes_values]

fig, axs = plt.subplots(1, len(passes_values), figsize=(15, 5))

for ax, img, p in zip(axs, imgs, passes_values):
    ax.imshow(img)
    ax.axis('off')
    ax.set_title(f'Passes={p}')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors
import matplotlib
import imageio.v2 as imageio
from PIL import Image

def get_colors_lospec(url):
    im = imageio.imread(url)
    rgb_list = (im/255)[0,:,0:3]
    float_list = list(np.linspace(0,1,len(rgb_list)))
    cdict = dict()
    for num, col in enumerate(['red', 'green', 'blue']):
        col_list = [[float_list[i], rgb_list[i][num], rgb_list[i][num]] for i in range(len(float_list))]
        cdict[col] = col_list
    cmp = mcolors.LinearSegmentedColormap('my_cmp', segmentdata=cdict, N=256)
    return cmp
```

```python
import matplotlib.pyplot as plt
import numpy as np
import base64
from io import BytesIO

# Function to convert the plot into a base64 string
def plot_to_base64():
    buf = BytesIO()
    plt.savefig(buf, format="png", bbox_inches='tight', pad_inches=0.1)
    plt.close()
    return base64.b64encode(buf.getvalue()).decode('utf-8')

# 1. Generate the plots and convert to Base64
x = np.linspace(0, 10, 100)

# Simple Line Plot
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
img_b64_1 = plot_to_base64()

# Scatter Plot
plt.scatter(x, np.random.randn(100), color='red')
img_b64_2 = plot_to_base64()

# Histogram
plt.hist(np.random.randn(1000), bins=30, color='blue')
img_b64_3 = plot_to_base64()

# Bar Plot
labels = ["A", "B", "C", "D"]
values = [10, 20, 15, 30]
plt.bar(labels, values, color='green')
img_b64_4 = plot_to_base64()

# 2. Generate the HTML content using f-strings
page = f"""
<!DOCTYPE html>
<html>
<head><title>Matplotlib Examples</title></head>
<body>
    <img src="data:image/png;base64,{img_b64_1}" alt="Line Plot" width="400">
    <img src="data:image/png;base64,{img_b64_2}" alt="Scatter Plot" width="400">
    <img src="data:image/png;base64,{img_b64_3}" alt="Histogram" width="400">
    <img src="data:image/png;base64,{img_b64_4}" alt="Bar Plot" width="400">
</body>
</html>
"""

# 3. Save the generated HTML content to a file
print(page, file=open("page_b64.html", 'w'))
```

```python
display(HTML(page))
```

```python
csv_path = "./file.csv"
```

```python
# Define dummy user data again
data = [
    ["Name", "Age", "Email"],
    ["John Doe", 25, "johndoe@example.com"],
    ["Jane Smith", 28, "jane.smith@example.com"],
    ["Alice Johnson", 22, "alice.johnson@example.com"],
    ["Bob Brown", 30, "bob.brown@example.com"]
]

# Write the data to a CSV file
with open(csv_path, mode='w', newline='') as file:
    writer = csv.writer(file)
    writer.writerows(data)

csv_path
```

```python
# Read the CSV file and extract the data
with open(csv_path, mode='r', newline='') as file:
    reader = csv.reader(file)
    csv_data = list(reader)

# Extract headers and rows
headers, rows = csv_data[0], csv_data[1:]

# Use f-string with loops to generate the HTML content
table_rows = "\n".join([f"<tr><td>{row[0]}</td><td>{row[1]}</td><td>{row[2]}</td></tr>" for row in rows])
html_content = f"""
<!DOCTYPE html>
<html>
<head>
    <title>Users List</title>
</head>
<body>
    <h1>List of Users</h1>
    <table border="1">
        <tr>
            <th>{headers[0]}</th>
            <th>{headers[1]}</th>
            <th>{headers[2]}</th>
        </tr>
        {table_rows}
    </table>
</body>
</html>
"""

# Save the generated HTML content to a file
html_file_path = "users_list.html"
print(html_content, file=open(html_file_path, 'w'))

html_file_path
```
