```
import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

# Define symbolic variables for SymPy
w_sym, b_sym, x_sym, y_sym = sp.symbols('w b x y')

# Predicted output (linear model) for symbolic representation
y_hat_sym = w_sym * x_sym + b_sym

# Define the loss function (MSE for a single example) symbolically
loss_sym = (y_sym - y_hat_sym)**2

# Calculate the partial derivatives (gradients) symbolically
partial_w_sym = sp.diff(loss_sym, w_sym)
partial_b_sym = sp.diff(loss_sym, b_sym)

# Display the symbolic gradients
print(f"Partial derivative of loss with respect to w: {partial_w_sym}")
print(f"Partial derivative of loss with respect to b: {partial_b_sym}")
```

```
# -------------------
# Practical implementation using NumPy with plots
# -------------------

# Sample data (input X and target Y)
X = np.array([1, 2, 3, 4])  # Input data (features)
Y = np.array([3, 5, 7, 9])  # Target data (labels)

# Initialize model parameters
w = 0.0  # Initial weight
b = 0.0  # Initial bias
learning_rate = 0.01  # Learning rate for gradient descent
epochs = 100  # Number of iterations

# Store the history of loss for plotting later
loss_history = []

# Gradient descent loop
for epoch in range(epochs):
    # Predicted values using the current model parameters
    Y_pred = w * X + b

    # Compute the gradients (derivatives of loss with respect to w and b)
    dw = -2 * np.dot(X, (Y - Y_pred)) / len(X)  # Gradient w.r.t w
    db = -2 * np.sum(Y - Y_pred) / len(X)       # Gradient w.r.t b

    # Update the model parameters using the gradients
    w = w - learning_rate * dw
    b = b - learning_rate * db

    # Calculate and store the loss (MSE)
    loss = np.mean((Y - Y_pred)**2)
    loss_history.append(loss)

# Plot Loss vs. Epochs
plt.figure(figsize=(10, 6))
plt.plot(range(epochs), loss_history, label="Loss over epochs")
plt.xlabel("Epochs")
plt.ylabel("Mean Squared Error (Loss)")
plt.title("Loss vs. Epochs")
plt.legend()
plt.grid(True)
plt.show()

# Plot the initial and final fits
plt.figure(figsize=(10, 6))

# Initial model (before training)
w_initial = 0.0
b_initial = 0.0
Y_pred_initial = w_initial * X + b_initial

# Final model (after training)
Y_pred_final = w * X + b

# Plot the data points
plt.scatter(X, Y, color="blue", label="Data Points")

# Plot the initial fit (before training)
plt.plot(X, Y_pred_initial, color="red", linestyle="--", label="Initial Fit")

# Plot the final fit (after training)
plt.plot(X, Y_pred_final, color="green", label="Final Fit")

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Linear Regression Fit Before and After Training")
plt.legend()
plt.grid(True)
plt.show()

# Print the final parameters after gradient descent
print(f"Final weight (w): {w}")
print(f"Final bias (b): {b}")
```

```
def softmax(x):
    s1 = torch.exp(x - torch.max(x,1)[0][:,None])
    s = s1 / s1.sum(1)[:,None]
    return s
```

```
def cross_entropy(outputs, labels):
    return -torch.sum(softmax(outputs).log()[range(outputs.size()[0]), labels.long()])/outputs.size()[0]
```

```
def randn_trunc(s): #Truncated Normal Random Numbers
    mu = 0
    sigma = 0.1
    R = stats.truncnorm((-2*sigma - mu) / sigma, (2*sigma - mu) / sigma, loc=mu, scale=sigma)
    return R.rvs(s)
```

```
def Truncated_Normal(size):

    u1 = torch.rand(size)*(1-np.exp(-2)) + np.exp(-2)
    u2 = torch.rand(size)
    z  = torch.sqrt(-2*torch.log(u1)) * torch.cos(2*np.pi*u2)

    return z
```

```
def acc(out,y):
    with torch.no_grad():
        return (torch.sum(torch.max(out,1)[1] == y).item())/y.shape[0]
```

```
def get_batch(mode):
    b = c.b
    if mode == "train":
        r = np.random.randint(X.shape[0]-b)
        x = X[r:r+b,:]
        y = Y[r:r+b]
    elif mode == "test":
        r = np.random.randint(X_test.shape[0]-b)
        x = X_test[r:r+b,:]
        y = Y_test[r:r+b]
    return x,y
```

```
def model(x,w):
    return x@w[0]
```

```
def make_plots():

    acc_train = acc(model(x,w),y)

    xt,yt = get_batch('test')

    acc_test = acc(model(xt,w),yt)

    wb.log({"acc_train": acc_train, "acc_test": acc_test})
```

```
for L in range(10):
    wb.init(project="Simple_Linear_Adam 2024");
    c = wb.config

    c.h = 0.01
    c.b = 1024
    c.epochs = 100

    w = [GPU(Truncated_Normal((784,10)))]

    optimizer = torch.optim.Adam(w, lr=c.h)

    for i in range(c.epochs):

        x,y = get_batch('train')

        loss = cross_entropy(softmax(model(x,w)),y)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

        wb.log({"loss": loss})

        make_plots()

        if i % 1000 == 0 : montage_plot((w[0].T).reshape(10,28,28).cpu().detach().numpy())
```
