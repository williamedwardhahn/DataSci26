```python
import numpy as np
import matplotlib.pyplot as plt
import base64
from io import BytesIO

# Convert the matplotlib plot to a base64 encoded image
def plot_to_base64():
    buf = BytesIO()
    plt.savefig(buf, format="png", bbox_inches='tight')
    plt.close()
    return base64.b64encode(buf.getvalue()).decode('utf-8')

def plot(x, y, title=None):
    fig, ax = plt.subplots()
    ax.plot(x, y)
    ax.axis('off')
    fig.set_size_inches(7, 7)
    plt.title(title)

html_content = "<html><body>"

x = np.linspace(0, 4 * np.pi, 1000)
frequencies = [1, 2, 3, 4]

for freq in frequencies:
    y_sin = np.sin(freq * x)
    y_cos = np.cos(freq * x)

    title = f"Sine and Cosine for frequency: {freq}"
    plot(x, y_sin, title)
    img_b64 = plot_to_base64()
    html_content += f"<h3>{title}</h3>"
    html_content += f'<img src="data:image/png;base64,{img_b64}" alt="{title}" />'


    plot(x, y_cos, title)
    img_b64 = plot_to_base64()
    html_content += f'<img src="data:image/png;base64,{img_b64}" alt="{title}" />'


html_content += "</body></html>"
print(html_content, file=open("sine_cosine_output.html",'w'))
```

```python
def histogram(x, title=None):
    fig, ax = plt.subplots()
    ax.hist(x, bins=30, color='green', alpha=0.7)
    ax.axis('off')
    fig.set_size_inches(7, 7)
    plt.title(title)

html_content = "<html><body>"

for i in range(3):
    x = np.random.randn(1000) + i*2 # Shifting mean

    title = f"Histogram {i+1}"
    histogram(x, title)
    img_b64 = plot_to_base64()
    html_content += f"<h3>{title}</h3>"
    html_content += f'<img src="data:image/png;base64,{img_b64}" alt="{title}" />'

html_content += "</body></html>"
print(html_content, file=open("histogram_output.html",'w'))
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
import torch
from skimage.util import montage

def plot(x):
    fig, ax = plt.subplots()
    im = ax.imshow(x, cmap = 'gray')
    ax.axis('off')
    fig.set_size_inches(18, 10)
    plt.show()

def smooth_filter(img_tensor_batch, target_color_tensor_batch, tolerance):
    # Calculate distances using broadcasting
    distances = torch.norm(img_tensor_batch * 255 - target_color_tensor_batch[:, None, None, :], dim=-1)

    # Normalize distances to [0, 1]
    normalized_distances = torch.clamp(distances / tolerance, 0, 1)

    # Calculate weights
    weights = (1 - normalized_distances).unsqueeze(-1)

    # Multiply image with weights using broadcasting
    filtered_img_tensor_batch = img_tensor_batch * weights

    return filtered_img_tensor_batch

# Read image from URL
url = "https://media.istockphoto.com/photos/colorful-flowers-collection-picture-id474140272"
# url = "https://www.superseeds.com/cdn/shop/products/cherry-bomb-tomato-organic-f1-hybrid-64-days-vegetables-pinetree-garden-seeds-731_800x.jpg"
image = io.imread(url) / 255.0

# Resize image
image = cv2.resize(image, dsize=(256, 256), interpolation=cv2.INTER_CUBIC)

# Create a batch of images
batch_size = 100
image_batch = np.repeat(image[np.newaxis, ...], batch_size, axis=0)

# Create random colors for each image in the batch
random_colors = (np.random.rand(batch_size, 3) * 255).astype(int)

# Convert images and colors to tensors
img_tensor_batch = torch.tensor(image_batch)
target_color_tensor_batch = torch.tensor(random_colors).float()

# Apply the modified functions
filtered = smooth_filter(img_tensor_batch, target_color_tensor_batch, 100.0).cpu().numpy()

```
