```python
def conv2(x,f):
    x2 = np.zeros(x.shape)
    for i in range(1,x.shape[0]-1):
        for j in range(1,x.shape[1]-1):

            x2[i,j] = f[0,0] * x[i-1,j-1] \
            +         f[0,1] * x[i-1,j] \
            +         f[0,2] * x[i-1,j+1] \
            +         f[1,0] * x[i,j-1] \
            +         f[1,1] * x[i,j] \
            +         f[1,2] * x[i,j+1] \
            +         f[2,0] * x[i+1,j-1] \
            +         f[2,1] * x[i+1,j] \
            +         f[2,2] * x[i+1,j+1]

    return x2
```

```python
import torch.nn.functional as F
import torch
import torch.nn as nn
import imageio as io
import numpy as np
import matplotlib.pyplot as plt
import torchvision
import torchvision.transforms as transforms
```

```python
def plot(x):
    fig, ax = plt.subplots()
    im = ax.imshow(x, cmap = 'gray')
    ax.axis('off')
    fig.set_size_inches(18, 10)
    plt.show()
```

```python
def scale1(x):
    return (x-np.min(x))/(np.max(x)-np.min(x))
```

```python
def process(image):
    image = np.mean(image,axis=2)
    image = scale1(image)
    image = image - np.mean(image)
    image = torch.from_numpy(image)
    image = image[None,:,:]
    return image
```

```python
def plot_map(z,color='red',thresh=3):
    z == torch.max(z)
    [y,x] = torch.where(torch.abs(z - torch.amax(z)) <  thresh)
    fig, ax = plt.subplots()
    im = ax.imshow(image[0,0], cmap = 'gray')
    ax.axis('off')
    fig.set_size_inches(18, 10)
    ax.scatter(x+7, y+7, c=color, s=40)
    plt.show()
    # print(x,y)
```

```python
image = io.imread("http://ian-albert.com/games/super_mario_bros_maps/mario-2-2.gif")

image = image[:,:,:]
image0 = image.copy()

mario = image[31:46,41:56,:]
jelly = image[152:167,312:327,:]
coin = image[185:200,224:239,:]
bug1 = image[105:120,1179:1194,:]
bug2 = image[80:95,1264:1279,:]
```

```python
plot(image)
plot(mario)
plot(jelly)
plot(coin)
plot(bug1)
plot(bug2)
```

```python
image = process(image)
mario = process(mario)
coin  = process(coin)
jelly = process(jelly)
bug1  = process(bug1)
bug2  = process(bug2)
```

```python
objects = torch.stack([mario,coin,jelly,bug1,bug2])
image   = image[None,:,:]
```

```python
maps = F.conv2d(image,objects)
```

```python
plot(image0)
plot_map(maps[0,0],'red',0.1)
plot_map(maps[0,1],'yellow')
plot_map(maps[0,2],'blue')
plot_map(maps[0,3],'green')
plot_map(maps[0,4],'magenta')
```
