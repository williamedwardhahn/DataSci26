```
import matplotlib.pyplot as plt
import numpy as np
from numpy.linalg import norm
import torch
```

```
n = 100
x = np.linspace(-10,10,n)
y = np.exp(-0.02*(x-4)**2) + 0.1*np.random.randn(n,)
```

```
plt.plot(x,y,'.')
```

```
x = x / norm(x) #Scale
y = y / norm(y)

r = np.random.permutation(n) #Shuffle
x = x[r]
y = y[r]

x1 = torch.tensor(x[:n//2]) #Split
x2 = torch.tensor(x[n//2:])
y1 = torch.tensor(y[None,:n//2])
y2 = torch.tensor(y[None,n//2:])

x1_b = torch.stack((x1,torch.ones(x1.shape[0]))) #Add Bias
x2_b = torch.stack((x2,torch.ones(x2.shape[0])))
```

```
h = 10
```

```
d = 1 # dim of data, function of single variable in this case
```

```
W1 = torch.rand(h,d+1).double()
```

```
H1 = torch.tanh(W1@x1_b)
```

```
W2 = (y1) @ H1.pinverse()
```

```
H2 = torch.tanh(W1@x2_b)
```

```
y3 = W2@H2
```

```
plt.plot(x2,y3[0],'bo',label="Prediction")
plt.plot(x1,y1[0],'ro',label="Training Data")
plt.legend(loc="upper left")
```

```
%%timeit
W2 = y1 @ torch.tanh(W1@x1_b).pinverse()
```

```
def ELM(x1,y1,x2,h):
    W1 = torch.rand(h,len(x1.shape)+1).double()
    return (y1 @ torch.tanh(W1@x1_b).pinverse())@torch.tanh(W1@x2_b)
```

```
y3 = ELM(x1,y1,x2,6)
```

```
plt.plot(x2,y3[0],'bo',label="Prediction")
plt.plot(x1,y1[0],'ro',label="Training Data")
plt.legend(loc="upper left")
```

Power of the function is hidden state size minus one (h-1), aka 2 is linear, 3 is a parabola, 4 is cubic, etc.

```
for i in [2, 3, 4, 5, 10, 50, 500]:
  y3 = ELM(x1,y1,x2,i)
  plt.plot(x2,y3[0],'bo',label="Prediction")
  plt.plot(x1,y1[0],'ro',label="Training Data")
  plt.legend(loc="upper left")
  plt.title(F"Hidden state size: {i}")
  plt.show()
```
