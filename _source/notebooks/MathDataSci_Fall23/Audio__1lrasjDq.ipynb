```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.io.wavfile import write
from IPython.display import Audio

# Parameters for the sine wave
start = 0
stop = 10
steps = 1000
frequency = 440  # Frequency of the sine wave

# Generate the time domain array
x = np.linspace(start, stop, steps)

# Create the sine wave signal
y = np.sin(2 * np.pi * frequency * x)

# Plot the sine wave
plt.plot(x, y)
plt.title('Sine Wave')
plt.xlabel('Time [s]')
plt.ylabel('Amplitude')
plt.grid(True)
plt.show()
```

```python
# Define the sample rate and the duration of the signal
sample_rate = 44100  # Sample rate in Hz
duration = stop - start  # Duration in seconds

# Generate a new time array for the audio signal
t = np.linspace(start, duration, int(sample_rate * duration), endpoint=False)

# Create the audio signal. We need to ensure it's in the audible range.
audio_signal = np.sin(2 * np.pi * frequency * t)

# Normalize the audio signal
audio_signal = audio_signal / np.max(np.abs(audio_signal))

# Play the audio
Audio(audio_signal, rate=sample_rate)
```
