```
import numpy as np
import matplotlib.pyplot as plt
from numpy import *
```

```
t = np.linspace(0,10,500)
data = 0.1*np.sin(4*t)
plt.plot(data)
```

```
m_train = 300
m_transient = 100
m_testing = 100

n_in = 1
n_res = 5
n_out = 1

a = 0.3
```

```
data += 0.01*np.random.normal(0,1,data.shape)
```

```
plt.plot(data[0:m_train],'.')
```

```
x = np.zeros((n_res))

X = np.zeros((1+n_in+n_res, m_train-m_transient))

Y = np.zeros((n_out,m_testing))
```

```
Wi = (np.random.rand(n_res,1+n_in)-0.5);    #Wi is the input weights

Wr = 0.01*(np.random.randn(n_res,n_res)-0.5);   #Wr is the reservoir weights, maps neuron activity to itself and other neurons
```

```
Wi.shape,Wr.shape    #Wi.shape and Wr.shape describe the matrix shape of the input data and of reservoir communications respectively
```

```
Yt = data[m_transient+1:m_train+1].T     #Yt is the training data
```

```
Wi
```

```
Wi.shape
```

```
x
```

```
Wr.shape
```

```
Wr@x
```

```
def relu(x):
  return np.fmax(0,x)
```

```
for t in range(m_train):

    u = data[t]

    x_new = Wi@r_[1,u] + Wr@x

    x_new = relu(x_new)

    x_old = x

    x = (1-a)*x_old + a*x_new

    if t > m_transient:
        X[:,t-m_transient] = r_[1,u,x]
```

```
plt.imshow(X[:,0:50])
```

```
Wo = Yt@np.linalg.pinv(X)   #Wo is the output layer
```

```
Wo.shape
```

```
Wo
```

```
u = data[m_train]

for t in range(m_testing):

    x_new = relu(Wi@r_[1,u] + Wr@x)

    x_old = x

    x = (1-a)*x_old + a*x_new

    u = Wo@r_[1,u,x]

    Y[:,t] = u
```

```
plt.plot(data[m_train+1:m_train+m_testing+1],'r.')
plt.plot(Y[0,:],'b.')
```

```
plt.plot(np.arange(m_testing),data[m_train-m_testing+1:m_train+1],'b.')
plt.plot(np.arange(m_testing,2*m_testing),data[m_train+1:m_train+m_testing+1],'r.')
plt.plot(np.arange(m_testing,2*m_testing),Y[0,:],'g*')
```
